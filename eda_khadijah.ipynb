```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_parquet("data/processed/tracts_merged.parquet")
```

```python
print(df.head())
```

```python
plt.hist(df["mental_distress"])
```

```python
plt.hist(df["depression"])
```

```python
print(df[(df["lahunv10share"].notna()) & (df["Urban"] == 0)])
```

```python
plt.scatter(df["LAPOP1_10"]/df["Pop2010"], df["depression"], s=5)
plt.xlabel("Share of people living far from the grocery store")
plt.ylabel("Depression")
plt.show()
```

```python
plt.scatter(df["lapop10share"], s=5)
plt.xlabel("Share of people living 10 miles away from the grocery store")
plt.ylabel("Depression")
plt.show()
```

```python
rural = df[df["Urban"] == 0]
urban = df[df["Urban"] == 1]

plt.scatter(rural["lapop10share"], rural["mental_distress"], s=5)
plt.xlabel("Share of people living far from the grocery store")
plt.ylabel("Depression")
plt.show()
```

```python
plt.scatter(urban["lapop1share"], urban["depression"], s=5)
plt.xlabel("Share of people living far from the grocery store")
plt.ylabel("Depression")
plt.show()
```
